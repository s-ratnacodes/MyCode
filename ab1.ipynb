```python
# final
import json
import random
import numpy as np
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch_geometric.datasets import Planetoid
from torch_geometric.nn import TAGConv
from torch_geometric.transforms import RandomNodeSplit
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")
plt.rcParams.update({"font.size": 12})

def set_seed(seed):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed(seed)
        torch.cuda.manual_seed_all(seed)
    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark = False

def get_data(dataset_name, split_type, seed):
    set_seed(seed)
    root = f"/tmp/{dataset_name}_{split_type}"
    if split_type == "fixed":
        dataset = Planetoid(root=root, name=dataset_name)
    else:
        transform = RandomNodeSplit(split="train_rest", num_val=0.10, num_test=0.10)
        dataset = Planetoid(root=root, name=dataset_name, transform=transform)
    return dataset, dataset[0]

class TAGIILayer(nn.Module):
    def __init__(self, in_channels, out_channels, x_org_channels, K, fixed_alpha):
        super().__init__()
        self.tagconv = TAGConv(in_channels, out_channels, K=K, normalize=True)
        self.need_proj = (x_org_channels != out_channels)
        if self.need_proj:
            self.proj = nn.Linear(x_org_channels, out_channels, bias=False)
        self.fixed_alpha = fixed_alpha

    def forward(self, x, x_org, edge_index):
        tag_out = self.tagconv(x, edge_index)
        residual = self.proj(x_org) if self.need_proj else x_org
        return (1.0 - self.fixed_alpha) * tag_out + self.fixed_alpha * residual

class TAGIINet(nn.Module):
    def __init__(self, num_features, num_classes, hidden, num_layers, K, dropout, fixed_alpha):
        super().__init__()
        dims = [num_features] + [hidden] * (num_layers - 1) + [num_classes]
        self.layers = nn.ModuleList([
            TAGIILayer(dims[i], dims[i+1], num_features, K, fixed_alpha)
            for i in range(num_layers)
        ])
        self.dropout = dropout

    def forward(self, x, edge_index):
        x_org = x
        for i, layer in enumerate(self.layers):
            x = F.dropout(x, p=self.dropout, training=self.training)
            x = layer(x, x_org, edge_index)
            if i < len(self.layers) - 1:
                x = F.relu(x)
        return F.log_softmax(x, dim=1)

def train_one_run(cfg, dataset, data, device, epochs=1000, patience=100):
    model = TAGIINet(
        dataset.num_node_features, dataset.num_classes,
        cfg["hidden"], cfg["num_layers"], cfg["K"],
        cfg["dropout"], cfg["fixed_alpha"]
    ).to(device)
    optimizer = torch.optim.Adam(model.parameters(), lr=cfg["lr"], weight_decay=cfg["weight_decay"])
    best_val, best_test = 0.0, 0.0
    patience_counter = 0

    for _ in range(epochs):
        model.train()
        optimizer.zero_grad()
        out = model(data.x, data.edge_index)
        loss = F.nll_loss(out[data.train_mask], data.y[data.train_mask])
        loss.backward()
        optimizer.step()

        model.eval()
        with torch.no_grad():
            pred = model(data.x, data.edge_index).argmax(dim=1)
            val_acc  = (pred[data.val_mask]  == data.y[data.val_mask]).float().mean().item()
            test_acc = (pred[data.test_mask] == data.y[data.test_mask]).float().mean().item()

        if val_acc > best_val:
            best_val, best_test = val_acc, test_acc
            patience_counter = 0
        else:
            patience_counter += 1
            if patience_counter >= patience:
                break
    return best_test

def run_seeds(cfg, dataset_name, split_type, seeds, device):
    accs = []
    for seed in seeds:
        dataset, data = get_data(dataset_name, split_type, seed)
        data = data.to(device)
        set_seed(seed)
        accs.append(train_one_run(cfg, dataset, data, device))
    return accs

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
SEEDS = list(range(10, 15))
FIXED_ALPHAS = [0.1, 0.5, 0.7]
K_VALUES = [3, 4, 5]

BASE_CFG = {
    "hidden": 64,
    "num_layers": 4,
    "dropout": 0.5,
    "lr": 0.01,
    "weight_decay": 5e-4,
}

DATASETS = ["Cora", "CiteSeer", "PubMed"]
SPLITS = ["fixed", "80:10:10"]

def run_all():
    results = {ds: {a: {k: [] for k in K_VALUES} for a in FIXED_ALPHAS} for ds in DATASETS}

    for ds in DATASETS:
        for split in SPLITS:
            print(f"\n{'='*60}")
            print(f"  {ds}  |  {split}")
            print('='*60)
            for alpha in FIXED_ALPHAS:
                for K in K_VALUES:
                    cfg = {**BASE_CFG, "fixed_alpha": alpha, "K": K}
                    print(f"  α={alpha:.1f}  K={K} ...", end=" ", flush=True)
                    accs = run_seeds(cfg, ds, split, SEEDS, DEVICE)
                    mean, std = np.mean(accs), np.std(accs)
                    print(f"{mean:.4f} ± {std:.4f}")
                    results[ds][alpha][K].extend(accs)

    save_dict = {
        ds: {str(a): {str(k): v for k, v in res_a.items()} for a, res_a in res.items()}
        for ds, res in results.items()
    }
    with open("tagii_alpha_K_ablation.json", "w") as f:
        json.dump(save_dict, f, indent=2)
    print("\nSaved → tagii_alpha_K_ablation.json")
    return results

def plot_grouped_bars(results, save_prefix="tagii_alpha_K_bars"):
    colors = ["#4C72B0", "#DD8452", "#55A868"]
    fig, axes = plt.subplots(1, 3, figsize=(14, 4.8), sharey=False)

    for ax, ds in zip(axes, DATASETS):
        x = np.arange(len(FIXED_ALPHAS))
        width = 0.25
        for i, (K, color) in enumerate(zip(K_VALUES, colors)):
            means = [np.mean(results[ds][a][K]) for a in FIXED_ALPHAS]
            stds  = [np.std(results[ds][a][K])  for a in FIXED_ALPHAS]
            offset = (i - 1) * width
            bars = ax.bar(x + offset, means, width, yerr=stds,
                          label=f"K={K}", color=color,
                          edgecolor="black", linewidth=0.7,
                          capsize=3.5, error_kw={"elinewidth": 1.2, "capthick": 1.2})
            for bar, m in zip(bars, means):
                ax.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.008,
                        f"{m:.3f}", ha="center", va="bottom", fontsize=7.5)

        ax.set_title(ds, fontsize=13, fontweight="semibold", pad=8)
        ax.set_xticks(x)
        ax.set_xticklabels([f"α={a}" for a in FIXED_ALPHAS])
        ax.set_xlabel("Fixed α")
        ax.grid(True, axis="y", linestyle="--", alpha=0.45)
        ax.set_axisbelow(True)
        ax.spines["top"].set_visible(False)
        ax.spines["right"].set_visible(False)
        all_vals = [np.mean(results[ds][a][k]) for a in FIXED_ALPHAS for k in K_VALUES]
        ax.set_ylim(min(all_vals) - 0.05, max(all_vals) + 0.05)

    axes[0].set_ylabel("Test Accuracy")
    axes[0].legend(title="Filter Order K", fontsize=10, title_fontsize=10,
                   frameon=True, edgecolor="#CCC", loc="lower right")
    fig.suptitle("TAGII: Ablation on Fixed α and Filter Order K", fontsize=14, fontweight="semibold", y=1.02)
    plt.tight_layout()
    plt.savefig(f"{save_prefix}.png", dpi=400, bbox_inches="tight", facecolor="white")
    plt.savefig(f"{save_prefix}.pdf", bbox_inches="tight")
    print(f"Saved → {save_prefix}.png + .pdf")
    plt.show()

def plot_multipanel_lines(results, save_prefix="tagii_alpha_K_lines"):
    colors = ["#4C72B0", "#DD8452", "#55A868"]
    markers = ["o", "s", "D"]
    fig, axes = plt.subplots(1, 3, figsize=(13, 4.5), sharey=False)

    for ax, ds in zip(axes, DATASETS):
        for K, color, marker in zip(K_VALUES, colors, markers):
            means = [np.mean(results[ds][a][K]) for a in FIXED_ALPHAS]
            stds  = [np.std(results[ds][a][K])  for a in FIXED_ALPHAS]
            ax.errorbar(FIXED_ALPHAS, means, yerr=stds,
                        color=color, marker=marker, markersize=8,
                        linewidth=2.1, capsize=4, capthick=1.4,
                        markeredgecolor="black", markeredgewidth=0.8,
                        label=f"K={K}")

        ax.set_title(ds, fontsize=13, fontweight="semibold", pad=8)
        ax.set_xticks(FIXED_ALPHAS)
        ax.set_xticklabels([f"{a}" for a in FIXED_ALPHAS])
        ax.set_xlabel("Fixed α")
        ax.grid(True, linestyle="--", alpha=0.45)
        ax.spines["top"].set_visible(False)
        ax.spines["right"].set_visible(False)
        all_vals = [np.mean(results[ds][a][k]) for a in FIXED_ALPHAS for k in K_VALUES]
        ax.set_ylim(min(all_vals) - 0.05, max(all_vals) + 0.05)

    axes[0].set_ylabel("Test Accuracy")
    axes[0].legend(title="Filter Order K", fontsize=10, title_fontsize=10,
                   frameon=True, edgecolor="#CCC")
    fig.suptitle("TAGII: Effect of Fixed α under Different K", fontsize=14, fontweight="semibold", y=1.02)
    plt.tight_layout()
    plt.savefig(f"{save_prefix}.png", dpi=400, bbox_inches="tight", facecolor="white")
    plt.savefig(f"{save_prefix}.pdf", bbox_inches="tight")
    print(f"Saved → {save_prefix}.png + .pdf")
    plt.show()

if __name__ == "__main__":
    print(f"Device : {DEVICE}")
    print(f"Seeds  : {SEEDS}")
    print(f"α      : {FIXED_ALPHAS}")
    print(f"K      : {K_VALUES}")

    results = run_all()

    print("\nSUMMARY")
    for ds in DATASETS:
        print(f"\n{ds}")
        for a in FIXED_ALPHAS:
            for k in K_VALUES:
                accs = results[ds][a][k]
                print(f"  α={a:.1f}  K={k} : {np.mean(accs):.4f} ± {np.std(accs):.4f}")

    plot_grouped_bars(results)
    plot_multipanel_lines(results)
    print("\nDone!")
```
